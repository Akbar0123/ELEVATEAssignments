```python
#sns for style purpose 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
```

```python
df = pd.read_csv("train.csv")
df.head()
```

```python
df = pd.read_csv("titanic/train.csv")
df.head()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
df.drop(columns=['Cabin'], inplace=True)  # Too many missing values
```

```python
sns.countplot(data=df, x='Survived')
plt.title("Survival Distribution")
plt.show()
```

```python
sns.histplot(df['Age'], kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.countplot(data=df, x='Sex', hue='Survived')
plt.title("Survival by Gender")
plt.show()
```

```python
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title("Survival by Passenger Class")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
numeric_df = df.select_dtypes(include=['float64', 'int64'])
plt.figure(figsize=(10,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df[['Survived','Age','Fare','Pclass']])
plt.show()
```

```python
sns.boxplot(data=df, x=df['Fare'])
plt.title("Fare Outliers")
plt.show()
```

```python
print("""
1. Females had a significantly higher survival rate.
2. First-class passengers survived more compared to third-class passengers.
3. Younger passengers (children) had better chances of survival.
4. Fare is positively correlated with survival probability.
5. Missing cabin data is too high to be useful.
6. Age distribution is slightly right-skewed.
""")
```
